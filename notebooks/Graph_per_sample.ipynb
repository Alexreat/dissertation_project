```python
import os
import networkx as nx
from cobra.io import load_matlab_model

def build_graph_from_mat_model(filepath):
    """
    Loads a COBRA model from a .mat file and builds a NetworkX graph.
    Nodes = metabolites
    Edges = participation in a reaction
    """
    model = load_matlab_model(filepath, variable_name="specificModel")
    G = nx.Graph()

    for rxn in model.reactions:
        mets = [met.id for met in rxn.metabolites]
        # Add edges between every pair of metabolites in the reaction
        for i in range(len(mets)):
            for j in range(i + 1, len(mets)):
                G.add_edge(mets[i], mets[j], reaction=rxn.id)

    return G


def load_all_graphs(folder_path):
    """
    Loads all .mat iMAT models from folder and builds graphs for each.
    Returns: dict {sample_name: G}
    """
    graphs = {}

    for fname in os.listdir(folder_path):
        if fname.endswith('.mat'):
            try:
                path = os.path.join(folder_path, fname)
                sample = fname.replace('_imat_vector.mat', '')
                G = build_graph_from_mat_model(path)
                graphs[sample] = G
                print(f"✅ Loaded graph for {sample}: {G.number_of_nodes()} nodes, {G.number_of_edges()} edges.")
            except Exception as e:
                print(f"❌ Failed to load {fname}: {e}")

    return graphs
```
